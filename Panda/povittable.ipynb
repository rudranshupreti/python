```python
import pandas as pd
import numpy as np  
df = pd.DataFrame({
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Department": ["HR", "IT", "HR", "IT", "Finance", "Finance", "HR", "IT"],
    "Gender": ["F", "M", "M", "M", "F", "M", "F", "M"],
    "Salary": [50000, 70000, 55000, 65000, 60000, 62000, 58000, 72000],
    "Experience": [2, 5, 3, 4, 6, 2, 4, 7]
})

print(df)
```

```python
pd.pivot_table(df,values='Salary',index="Gender", columns="Department")
```

```python
#crosstab
pd.crosstab(df["Gender"],df["Department"])
```
