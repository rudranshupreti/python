```python
import pandas as pd

df1 = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Department": ["HR", "IT", "Finance", "Sales"]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 105],
    "Salary": [50000, 65000, 70000, 55000]
})

print(df2)
```

```python
pd.merge(df1,df2, on="Emp_ID",how='inner')
```

```python
pd.merge(df1,df2, on="Emp_ID",how='outer')
```

```python
pd.merge(df1,df2, on="Emp_ID",how='left')
```

```python
pd.merge(df1,df2, on="Emp_ID",how='right')
```

```python
pd.concat([df1,df2],axis= 0)
```

```python
pd.concat([df1,df2],axis= 1)
```

```python
df1.join(df2)
```
