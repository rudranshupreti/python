```python
# Sorting a NumPy array based on a second column
import numpy as np
sam = np.array([[34, 43, 73], [82, 22, 12], [53, 94, 66]])
sort_arr = sam[sam[:, 1].argsort()]

print(sort_arr)
```

```python
# Delete the second column from a given array and insert the following new column in its place.
import numpy as np

sampleArray = np.array([[34,43,73],[82,22,12],[53,94,66]])
newColumnToAdd = np.array([10, 10, 10])
dt = np.delete(sampleArray,1,axis=1)
ins = np.insert(dt,1,newColumnToAdd,axis=1)
print(sampleArray)
print(dt)
print(ins)
```

```python
# Exercise 29: Swap column 1 and 2 in a 2D array
import numpy as np
arr = np.arange(9).reshape(3, 3)
# The columns are at index 0, 1, 2. We want to rearrange to 0, 2, 1
print(arr)
arr[:,[1,2]] = arr[:,[2,1]]
print(arr)
```

```python
# Exercise 35: Sort the rows of a 2D array based on the values of the second column
import numpy as np
arr = np.array([[8, 4, 1],
                [5, 2, 7],
                [6, 9, 3]])
srt = arr[:,1].argsort()
print(arr[srt])
```

```python
# Exercise 42: Count the number of occurrences of each unique element
import numpy as np
arr = np.array([2, 3, 2, 5, 3, 3, 2, 5])
vals , counts = np.unique(arr,return_counts=True)
print(vals,counts)
```

```python
# Exercise 43: Find the intersection and union of two arrays
import numpy as np

a = np.array([1, 2, 3, 5, 7])
b = np.array([3, 4, 5, 6, 7])

int = np.intersect1d(a,b)
un = np.union1d(a,b)
print(int , un)
```
